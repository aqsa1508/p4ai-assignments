```python
                                                            #QUESTION NO 1
stu=[]
for i in range(0,5):
    print("enter details of sudent\n")
    name=input("enter name ")
    r_n=input("enter roll ")
    python=int(input("enter marks of python"))
    ai=int(input("enter marks of AI"))
    maths=int(input("enter marks of maths"))
    student={
        "name":name,
        "roll_no":r_n,
        "sub_marks":{
            "pyt":python,
            "AI":ai,
            "mathematics":maths,
        }
    }
    stu.append(student)

def calculation(student):
    p= student["sub_marks"]["pyt"]
    a= student["sub_marks"]["AI"]
    m= student["sub_marks"]["mathematics"]
    total = p+ a+ m

    perc=(total/300)*100
    if(perc>=80 and perc<=100):
        grade="A"
    elif(perc<80 and perc >=70):
        grade="B"
    elif(perc<70 and perc >=60):
        grade="c"
    elif(perc<60 and perc >=50):
        grade="D"
    elif(perc<50 and perc >=0):
        grade="F"
    else:
        print("invalid marks ")
    print("name = ",student["name"])
    print("total marks = ",total)
    print("percentage = ",perc)
    print("grade = ",grade)
    student["total"] = total
    student["percentage"] = perc
    student["grade"] = grade
    
    f=open("result.txt","a")
    f.write("Name = " + student["name"] + "\n")
    f.write("Total marks = " + str(total) + "\n")
    f.write("Percentage = " + str(perc) + "%\n")
    f.write("Grade = " + grade + "\n")
    f.close 
for student in stu:
    calculation(student)
highest = stu[0]

for student in stu:
    if student["percentage"] > highest["percentage"]:
        highest = student

print("\n STUDENT WITH HIGHEST PERCENTAGE ")
print("Name =", highest["name"])
print("Roll No =", highest["roll_no"])
print("Percentage =", highest["percentage"])
print("Grade =", highest["grade"])
```

```python
                                                             #QUESTION NO 2
#FOR TOTAL NUMBER OF WORDS 
s=input("write a sentence")
tw=s.split()
a=str(len(tw))
print("total number of words are ",len(tw))

#FOE TOTAL NUMBER OF CHARACTERS
count=0
for i in s:
    count+=1
b=str(count)
print("total number of characters are : ",count)

#FOR LONGEST WORD
maximum=tw[0]
j=0
for i in tw:
    if(len(tw[j])>len(maximum)):
        maximum=tw[j]
    j+=1
c=str(maximum)
print("longest word is",maximum)

#FOR SHORTEST WORD
minimum=tw[0]
j=0
for i in tw:
    if(len(tw[j])<len(minimum)):
        minimum=tw[j]
    j+=1
d=str(minimum)
print("shortest word is",minimum)

#FOR TOTAL NUMBER OF VOWELS
count=0
for i in s:
    if(i in["a","e","i","o","u","A","E","I","O","U"]):
        count+=1
e=str(count)
print("total number of vowels are : ",count)

#FOR SET OF UNIQUE WORDS
l=list()
for i in tw:
    if i not in l:
        l.append(i)
g=str(l)
print("set of unique words are:",l)

#FOR SORTING IN ALPHABETICAL ORDER
tw.sort()
h=str(tw)
print("after sorting in alphabetical order ",tw)

#FILE HANDLING
f=open("analysis.txt","a")
f.write(s+"\n")
f.write("total words = " + a + "\n")
f.write("total characters = " + b + "\n")
f.write("total longest word = " + c + "\n")
f.write("total shortest word = " + d + "\n")
f.write("total number of vowels = " + e + "\n")
f.write("st of unique word = " + g + "\n")
f.write("sorted in alphabetical order = " + h + "\n")
f.close()
```

```python
                                                       #QUESTION NO 3
product=[]
for i in range(0,5):
    print("enetr product detail")
    p_name=input("enter name: ")
    p_price=int(input("enter price : "))
    quan=int(input("enter quantity:"))
    prod={
    "name":p_name,
    "price":p_price,
    "qunatity":quan
    }
    product.append(prod)
def calculation(prod):
    p=prod["price"]
    q=prod["qunatity"]
    prod_price=p * q
    prod["total_price"]= prod_price    
    return prod_price
bill=0
for prod in product:
    bill+=calculation(prod)
if(bill>=10000):
    dis=bill*0.10
else:
    dis=0
total_bill=bill-dis
print("total bill = ",total_bill)
highest=product[0]
for prod in product:
    if(prod["price"]>highest["price"]):
       highest=prod
print("most expensive product ")
print("name=",highest["name"])
print("price=",highest["price"])
print("qunatity=",highest["qunatity"])
print("total price=",highest["total_price"])
for prod in product:
    f=open("bill.txt","a")
    f.write("name = :" + prod["name"] + "\n")
    f.write("price = :" + str(prod["price"]) + "\n")
    f.write("quantity = :" + str(prod["qunatity"]) + "\n")
    f.write("total price = :" + str(prod["total_price"]) + "\n")
    f.close()
f=open("bill.txt","a")
f.write("total bill = :" + str(total_bill) + "\n")
f.close()
```
